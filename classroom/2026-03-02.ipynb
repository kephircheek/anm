```python
tables = pd.read_html(
    van_der_vaals_eq_wiki_page, 
    decimal=",",
    thousands=None,
)
df = tables[0]
df.head()
```

```python
df["Вещество"].str.rsplit(" ", n=1, expand=True)
```

```python
df_with_formula = df.copy()
df_with_formula[["Название", "Формула"]] = \
    df["Вещество"].str.rsplit(" ", n=1, expand=True)
df_with_formula.drop(columns=["Вещество"], inplace=True)
# df_with_formula = df_with_formula.drop(columns=["Вещество"])

df_with_formula.head()
```

```python
list(df_with_formula.columns)[2:] \
    + list(df_with_formula.columns)[:2]
```
