```python
%pip install lxml
url = (
    "https://ru.wikipedia.org/"
    "api/rest_v1/page/html/"
    "Уравнение_Ван-дер-Ваальса"
)
r = requests.get(url)

import pandas as pd
df = pd.read_html(
    r.text, decimal=",", thousands=None
)[0]

df_substance = df["Вещество"] \
    .str.rsplit(" ", n=1, expand=True) \
    .rename(columns={0: "Name", 1: "Formula"})
df_substance
```

```python
elements = {}
element_, element, count = "", "", ""
is_alpha, is_count = True, False
for l in "C2H2":
    if l.isalpha():
        if is_count:
            elements[element_] = count
            count = ""
        is_alpha, is_count = True, False
        element += l
    else:
        if is_alpha:
            element_ = element
            element = "" 
        is_alpha, is_count = False, True
        count += l

elements
```
