## Unzip

```python
x_arr, y_arr = [1, 3, 2], [11, 13, 12]
sorted_data = sorted(list(zip(x_arr, y_arr)))
sorted_data
```

```python
plt.plot(sorted_data, ".-")
```

```python
sorted_data
```

```python
list(zip(sorted_data))
```

```python
list(zip(sorted_data[0], sorted_data[1], sorted_data[2]))
```

```python
list(zip(*sorted_data))
```

```python
x_arr_s, y_arr_s = list(zip(*sorted_data))
```

```python
plt.plot(x_arr_s, y_arr_s)
```

## Pandas

```python
import pandas as pd
```

```python
df = pd.DataFrame([[1, 3, 2], [11, 13, 12]])
df
```

```python
df = pd.DataFrame([[1, 3, 2], [11, 13, 12]]).T
df
```

```python
df = df.rename(columns={0: "x", 1: "y"})
df
```

```python
df = df.sort_values(by="x")
df
```

```python
df.set_index("x")
```

```python
df.set_index("x").plot()
```

```python
pd.DataFrame([[1, 3, 2], [11, 13, 12]]) \
    .T \
    .rename(columns={0: "x", 1: "y"}) \
    .sort_values(by="x") \
    .set_index("x") \
    .plot()
```
