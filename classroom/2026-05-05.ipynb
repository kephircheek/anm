```python
import chemparse

def parse_formula(formula):
    return tuple(sorted(chemparse.parse_formula(formula).items()))
```

```python
vdv_consts["Parsed_Formula"] = vdv_consts.Formula.map(parse_formula)
compounds["parsed_formula"] = compounds.molecular_formula.map(parse_formula)

vdv_consts_rich = vdv_consts.join(
    compounds.groupby("parsed_formula").first(), 
    on="Parsed_Formula",
    how="inner",
)
    
vdv_consts_rich.shape
```
