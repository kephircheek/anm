```python
def parse_molecular_formula(formula):
    return tuple(sorted(chemparse.parse_formula(formula).items()))


compounds["parsed_formula"] = compounds.molecular_formula.map(
    parse_molecular_formula
)
compounds.head(1)
```

```python
vdv_const["parsed_formula"] = vdv_const.Formula.map(
    parse_molecular_formula
)
vdv_const.head(1)
```

```python
vdv_const_rich = vdv_const.set_index("parsed_formula").join(
    compounds.groupby("parsed_formula").first(),
    how="inner"
)

vdv_const_rich.shape
```

```python
vdv_const_rich.head(1)
```

```python
vdv_const_features = vdv_const_rich[
  ["Formula", "Const_a", "Const_b", "Mol_weight", "Volume" , "bonds"]
]
vdv_const_features.set_index("Formula", inplace=True)
vdv_const_features.head(1)
```

```python
import json
vdv_const_features["n_bonds"] = vdv_const_features.bonds.map(
    lambda bonds: sum(
        b["order"] for b in json.loads(bonds.replace("'", '"'))
    )
)
vdv_const_features.head(1)
```

```python
vdv_const_features = vdv_const_features.drop(columns=["bonds"])
```

```python
vdv_const_features.head()
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=5)
pca.fit(vdv_const_features)
pca.explained_variance_ratio_
```
